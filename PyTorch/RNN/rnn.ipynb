```python
import pandas as pd
df=pd.read_csv(r"E:\Downloads\training_sheets\100_Unique_QA_Dataset.csv")
df.head()
```

```python
# Tokenize
def tokenize(text):
    text=text.lower()
    text=text.replace('?',"")
    text=text.replace("'","")
    return text.split()
```

```python
# Vocabulary
vocab={'<UNK>':0}
def build_vocab(row):
    tokenized_question=tokenize(row['question'])
    tokenized_answer=tokenize(row['answer'])
    merged_tokens=tokenized_question+tokenized_answer

    for token in merged_tokens:
        if token not in vocab:
            vocab[token]=len(vocab)
```

```python
df.apply(build_vocab,axis=1)
len(vocab)
```

```python
vocab
```

```python
# convert words to numerical indicies
def text_to_indicies(text,vocab):
    indexed_text=[]
    for token in tokenize(text):
        if token in vocab:
            indexed_text.append(vocab[token])
        else:
            indexed_text.append(vocab['<UNK>'])
    return indexed_text
```

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class QADataset(Dataset):
    def __init__(self,df,vocab):
        self.df=df
        self.vocab=vocab

    def __len__(self):
        return self.df.shape[0]

    def __getitem__(self, index):
        num_question=text_to_indicies(self.df.iloc[index]['question'],self.vocab)
        num_answer=text_to_indicies(self.df.iloc[index]['answer'],self.vocab)

        return torch.tensor(num_question),torch.tensor(num_answer) 
```

```python
dataset=QADataset(df,vocab)
dataset[0]
```

```python
dataloader=DataLoader(dataset,batch_size=1,shuffle=True)
```

```python
import torch.nn as nn
```

```python
class SimpleRNN(nn.Module):
    def __init__(self,vocab_size):
        super().__init__()
        self.embedding=nn.Embedding(vocab_size,embedding_dim=50)
        self.rnn=nn.RNN(50,64,batch_first=True)
        self.fc=nn.Linear(64,vocab_size)

    def forward(self,question):
        embedded_ques=self.embedding(question)
        hidden,final=self.rnn(embedded_ques)
        output=self.fc(final.squeeze(0))

        return output

# We did not use Sequential because the RNN gives even intermediate outputs and final ouput but we need only final output so in sequential we cannot do that so we should do manually
```

```python
lr=0.001
epochs=20
```

```python
model=SimpleRNN(len(vocab))
```

```python
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=lr,)
```

```python
# training loop
for epoch in range(epochs):
    total_loss=0
    for question,answer in dataloader:

        #forward
        output=model.forward(question)

        #loss
        loss=criterion(output,answer[0])

        # gradients
        optimizer.zero_grad()
        loss.backward()

        #update grads
        optimizer.step()

        total_loss+=loss.item()

    print(f"Epoch:{epoch+1},Loss:{total_loss:4f}")
```

```python
def predict(model,question,threshold=0.5):
    # convert question to numbers
    num_ques=text_to_indicies(question,vocab)

    #tensor
    ques_tensor=torch.tensor(num_ques).unsqueeze(0)

    # send to model
    output=model(ques_tensor)

    # convert logits to probs
    probs=torch.nn.functional.softmax(output,dim=1)

    # find idx of max prob
    value,idx=torch.max(probs,dim=1)

    if value<threshold:
        print("I don't know")
    else:
        print(list(vocab.keys())[idx])
    
```

```python
predict(model,"What is the Largest Planet in our solar system?")
```
