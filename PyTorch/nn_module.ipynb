```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
df.head()
```

```python
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2)
```

```python
scaler=StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

```python
y_train
```

```python
# Label Encoder
encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

```python
# numpy arrays to pytorch tensors
X_train_tensor=torch.from_numpy(X_train).float()
X_test_tensor=torch.from_numpy(X_test).float()
y_train_tensor=torch.from_numpy(y_train).float().view(-1,1)
y_test_tensor=torch.from_numpy(y_test).float().view(-1,1)
```

```python
y_train_tensor.shape
```

```python
# creating model
class Model(nn.Module):
    def __init__(self,num_features):
        super().__init__()
        self.network=nn.Sequential(
            nn.Linear(num_features,3),  #------> Takes input features and gives 3 outputs(3 nodes)
            nn.ReLU(),
            nn.Linear(3,1),              #-----> Takes 3 input features and gives 1 output(1 node)
            nn.Sigmoid()
        )

    def forward(self,features):
        out=self.network(features)
        return out
```

```python
# params
learning_rate=0.1
epochs=25
```

```python
loss_function=nn.BCELoss()
```

```python
# Training Pipeline

#create model
model=Model(X_train_tensor.shape[1])

#define the optimizer
optimizer=torch.optim.SGD(model.parameters(),lr=learning_rate)

# define in loop(epochs)
for epoch in range(epochs):

    #forward pass
    y_pred=model(X_train_tensor)
    
    #loss calculation
    loss=loss_function(y_pred,y_train_tensor)

    # clear gradients
    optimizer.zero_grad()

    #backward pass
    loss.backward()

    # weights and bais update
    optimizer.step()

    #prit loss 
    print(f"Epoch:{epoch+1},loss:{loss.item()}")
```

```python
# Evaluation
with torch.no_grad():
    y_pred=model.forward(X_test_tensor)
    y_pred=(y_pred>0.62).float()
    acc=(y_pred==y_test_tensor).float().mean()
    print(f"Accuracy:{acc.item()}")
```
