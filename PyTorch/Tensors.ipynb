Matrix Operations

```python
f=torch.randint(size=(2,3),low=0,high=10)
g=torch.randint(size=(3,2),low=0,high=10)

print(f)
print(g)
```

```python
# matrix mul
torch.matmul(f,g)
```

```python
#dot product
vec1=torch.tensor([1,2])
vec2=torch.tensor([3,4])

torch.dot(vec1,vec2)
```

```python
#Transpose
print(f)
print(torch.transpose(f,0,1))
```

```python
#det
h=torch.randint(size=(3,3),low=0,high=10,dtype=torch.float)
print(h)
torch.det(h)
```

```python
#inverse
torch.inverse(h)
```

Comparision operations

```python
i=torch.randint(size=(2,3),low=0,high=10)
j=torch.randint(size=(2,3),low=0,high=10)

print(i)
print(j)
```

```python
#greater than
print(i>j)

#less than
print(i<j)

#not equal to
print(i!=j)

# greater than equal to
print(i>=j)

#less than or equal to
print(i<=j)
```

Special Functions

```python
k= torch.randint(size=(2,3),low=0,high=10,dtype=torch.float)
k
```

```python
#log
print(torch.log(k))

#exp
print(torch.exp(k))

#sqrt
print(torch.sqrt(k))

#sigmoid
print(torch.sigmoid(k))

#softmax
print(torch.softmax(k,dim=0))

#relu
print(torch.relu(k))
```

Inplace operation

```python
m=torch.rand(2,3)
n=torch.rand(2,3)

print(m)
print(n)
```

```python
m.add_(n)  # now result in stored in m
```

```python
m.relu_() # applies relu and stored in m
```
