```python
# small neural network type
#inputs
x=torch.tensor(6.7) #input
y=torch.tensor(0.0) # true label

w=torch.tensor(1.0) # weight
b=torch.tensor(0.0) # bias
```

```python
# binary cross entropy
def binary_cross_entropy_loss(pred,tar):
    epsilon=1e-8  # to prevent log(0)
    pred=torch.clamp(pred,epsilon,1-epsilon)
    return -(tar*torch.log(pred)+(1-tar)*torch.log(1-pred))
```

```python
# forward pass
z= w*x + b
y_pred=torch.sigmoid(z)

loss=binary_cross_entropy_loss(y_pred,y)
```

```python
# Derivatives:
# dL/d(y_pred):
dloss_dy_pred=(y_pred-y)/(y_pred*(1-y_pred))

# dy_pred/dz
dy_pred_dz=y_pred*(1-y_pred)

# dz/dw and dz/db
dz_dw=x
dz_db=1

dL_dw=dloss_dy_pred*dy_pred_dz*dz_dw
dL_db=dloss_dy_pred*dy_pred_dz*dz_db
```

```python
print(f"Manual Gradient of loss w.r.t weight(dw):{dL_dw}")
print(f"Manual Gradient of loss w.r.t bias(db):{dL_db}")
```

```python
x=torch.tensor(6.7)
y=torch.tensor(0.0)

w=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)

z = w*x + b
y_pred=torch.sigmoid(z)
loss=binary_cross_entropy_loss(y_pred,y)

loss.backward()

print(w.grad)
print(b.grad)
```

Clearing grad

```python
# if u repaet the process of calculating gradients u will accumulate the grads and keep on adding only so to avoid thta we use the 'clear_grad'
x=torch.tensor(2.0,requires_grad=True)
y=x**2
y.backward()
x.grad
```

How to stop gradient tracking

```python
# After the training at inference u no need of gradients
# 1. requires_grad_(False)
# 2. detach()
# 3. torch.no_grad()

x=torch.tensor(2.0,requires_grad=True)
y=x**2
x.requires_grad_(False)
print(x)
print(y)

x=torch.tensor(2.0,requires_grad=True)
z=x.detach()
print(x)
print(z) # same values of x but not able to calculate grad

x=torch.tensor(2.0,requires_grad=True)
print(x)
with torch.no_grad():
    y=x**2
print(y)
```
