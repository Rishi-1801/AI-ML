### Prompt Template Text Summarization

```python
from langchain_core.prompts import PromptTemplate
template="""
Write a summary of the following speech:
Speech:{speech}
Translate the precise summary  to {language}
"""
prompt=PromptTemplate(
    template=template,
    input_variables=['speech','language']
)
prompt
```

```python
chain= prompt|model
chain
```

```python
chain.invoke({'speech':speech,"language":"hindi"}).content
```

### StuffDocumentChain Text Summarization

```python
from langchain_classic.chains.summarize import load_summarize_chain
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader=PyPDFLoader(r"D:\AI\LangChain\Project\research_papers\apjspeech.pdf")
docs=loader.load()
docs
```

```python
template="""
Write a concise and short summary of the following speech
Speech:{text}
"""
prompt=PromptTemplate(
    template=template,
    input_variables=['text']
)
```

```python
chain=load_summarize_chain(model,chain_type='stuff',prompt=prompt,verbose=True)
# it combines all the docs to one text and pass to prompt and next to llm forr summarization
chain.invoke(docs)
```

### Map Reduce to Summarize Large Documnets

```python
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=2000,chunk_overlap=100)
chunks=text_splitter.split_documents(docs)
chunks
```

```python
chunks_template="""
Please summarize the blow speech:
Speech:{text}
Summary:
"""
map_prompt=PromptTemplate(template=chunks_template,
                          input_variables=['text'])

# Here we will get a list of summaries--> 1 summary for each chunk
```

```python
final_prompt="""
Provide the final Summary of the entire speech with these impoert points.
Add a Motivation Title, Start th eprecise summary with an introduction and provide the summary in number points for the speech.
Speech:{text}
"""
final_prompt_template=PromptTemplate(
    template=final_prompt,
    input_variables=['text']
)
```

```python
summary_chain=load_summarize_chain(
    llm=model,
    chain_type='map_reduce',
    map_prompt=map_prompt,
    combine_prompt=final_prompt_template,
    verbose=True
)
output=summary_chain.invoke(chunks)
output
```

### Refine Chain For Summarization

```python
chain=load_summarize_chain(
    llm=model,
    chain_type='refine',
    verbose=True
)
output=chain.invoke(chunks)
output
```
